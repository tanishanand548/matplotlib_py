```python
# import matplotlib

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,12,15,13,25]

plt.plot(x,y)
plt.show()
```

```python
plt.Figure(figsize=(8,4))
plt.plot(x,y, color = 'green', marker='o', linestyle = 'dashed', linewidth = 2, markersize = 12)
plt.title("Figure")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [10,12,15,13,25]
y2 = [20,22,25,23,35]

plt.plot(x,y1, label = 'Sales 2024')
plt.plot(x,y2, label = 'Sales 2025')

plt.title("YoY Sales")
plt.xlabel('Month')
plt.ylabel('Sales')

plt.legend()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,12,15,13,25]

plt.bar(x,y)
plt.show()
```

```python
import random
data = [random.randint(1,10) for _ in range(100)]

plt.hist(data, bins=5)
plt.title("Histogram")
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D', 'E']
sales = [10,20,30,40,50]

plt.pie(sales, labels=categories, autopct='%1.1f%%', startangle=45)
plt.title("Pie Chart")
plt.show()
```

```python
y1 = [10,12,15,13,25]
y2 = [20,22,25,23,35]

plt.scatter(y1,y2)
plt.title("Scatter Chart")
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D', 'E']
sales = [10,20,30,40,50]

y1 = [10,12,15,13,25]
y2 = [20,22,25,23,35]

plt.subplot(1,2,1)
plt.bar(categories, sales)
plt.title("Bar Chart")

plt.subplot(1,2,2)
plt.scatter(y1,y2, color='yellow')
plt.title("Scatter Chart")
```

```python
# import pandas
import pandas as pd

data = {
    'Month' : ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Sales' : [12000, 10000, 11000, 15000, 25000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'], df['Sales'])
plt.title("Matplotlib with pandas")
plt.show()
```

```python
# save the image

plt.bar(df['Month'], df['Sales'])
plt.title("Matplotlib with pandas")
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('Monthly_sales_from_df.png')
plt.show()
```
